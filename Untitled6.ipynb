```python
from google.colab import files
from ultralytics import YOLO
from transformers import pipeline
from PIL import Image
import numpy as np, cv2, pandas as pd, matplotlib.pyplot as plt
from IPython.display import display

seg = YOLO("yolo26n-seg.pt")  # segmentation model
depth_model = pipeline("depth-estimation", model="depth-anything/Depth-Anything-V2-Small-hf")

path = list(files.upload())[0]
r = seg.predict(path, conf=0.4, verbose=False)[0]
if r.masks is None:
    raise SystemExit("No objects found")

img = cv2.cvtColor(r.orig_img, cv2.COLOR_BGR2RGB)
H, W = img.shape[:2]
depth = np.array(depth_model(Image.fromarray(img))["depth"].resize((W, H)), dtype=np.float32)

# Build the data table: one row per detected object
rows = []
masks = {}
for k in range(len(r.boxes)):
    mask = np.zeros((H, W), np.uint8)
    cv2.fillPoly(mask, [r.masks.xy[k].astype(np.int32)], 1)
    ys, xs = np.where(mask)
    if len(xs) == 0:
        continue
    masks[k] = mask

    d = depth[ys, xs]
    x1, y1, x2, y2 = [int(v) for v in r.boxes.xyxy[k].tolist()]
    rows.append({
        "id": k + 1,
        "object": seg.names[int(r.boxes.cls[k])],
        "confidence_%": round(float(r.boxes.conf[k]) * 100, 1),
        "box_width_px": x2 - x1,
        "box_height_px": y2 - y1,
        "mask_area_px": int(mask.sum()),
        "area_%_of_image": round(mask.sum() / (H * W) * 100, 2),
        "depth_mean": round(float(d.mean()), 1),
        "depth_min": round(float(d.min()), 1),
        "depth_max": round(float(d.max()), 1),
        # 100 = as close as the nearest point in the image, 0 = farthest
        "closeness_0_100": round(float(d.mean()) / float(depth.max()) * 100, 1),
    })

df = pd.DataFrame(rows).sort_values("closeness_0_100", ascending=False).reset_index(drop=True)
print(f"Image size: {W}x{H}, objects found: {len(df)}")
display(df.style
        .background_gradient(subset=["closeness_0_100"], cmap="Blues")
        .background_gradient(subset=["confidence_%"], cmap="Greens")
        .set_caption("Detected objects (sorted from nearest to farthest)"))
df.to_csv("objects.csv", index=False)
files.download("objects.csv")

# 2D cutout and 3D point cloud for the 3 most confident objects
for k in r.boxes.conf.argsort(descending=True)[:3]:
    k = int(k)
    if k not in masks:
        continue
    mask = masks[k]
    ys, xs = np.where(mask)

    cut = np.dstack([img, mask * 255])[ys.min():ys.max() + 1, xs.min():xs.max() + 1]

    if len(xs) > 5000:
        i = np.random.choice(len(xs), 5000, replace=False)
        xs, ys = xs[i], ys[i]

    fig = plt.figure(figsize=(12, 5))
    ax = fig.add_subplot(1, 2, 1)
    ax.imshow(cut)
    ax.axis("off")
    ax.set_title(f"#{k + 1} {seg.names[int(r.boxes.cls[k])]} {float(r.boxes.conf[k]) * 100:.0f}%")

    ax = fig.add_subplot(1, 2, 2, projection="3d")
    ax.scatter(xs, depth[ys, xs], -ys, c=img[ys, xs] / 255, s=3)
    ax.view_init(15, -45)
    ax.set_title("3D point cloud")
    plt.show()
```

```python
import matplotlib.pyplot as plt

labels = [f"#{i} {o}" for i, o in zip(df["id"], df["object"])]

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Graph 1: how close each object is
axes[0].barh(labels, df["closeness_0_100"], color="steelblue")
axes[0].set_title("Closeness (higher = nearer)", fontweight="bold")
axes[0].set_xlabel("closeness_0_100")
axes[0].invert_yaxis()

# Graph 2: model confidence
axes[1].barh(labels, df["confidence_%"], color="seagreen")
axes[1].set_title("Confidence (%)", fontweight="bold")
axes[1].set_xlabel("confidence_%")
axes[1].set_xlim(0, 100)
axes[1].invert_yaxis()

# Graph 3: size vs depth (bubble = object size)
sc = axes[2].scatter(df["depth_mean"], df["area_%_of_image"],
                     s=df["mask_area_px"] / 20 + 50,
                     c=df["confidence_%"], cmap="viridis", alpha=0.8)
for _, row in df.iterrows():
    axes[2].annotate(f"#{row['id']} {row['object']}",
                     (row["depth_mean"], row["area_%_of_image"]),
                     textcoords="offset points", xytext=(8, 6))
axes[2].set_title("Depth vs size (color = confidence)", fontweight="bold")
axes[2].set_xlabel("depth_mean (higher = nearer)")
axes[2].set_ylabel("area_% of image")
plt.colorbar(sc, ax=axes[2], label="confidence %")

plt.tight_layout()
plt.show()
```
